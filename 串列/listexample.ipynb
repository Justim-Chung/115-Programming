# 簡介

串列 (list) 是一個可以儲存多個元素 (整數，浮點數，或字串) 的容器。

串列中每個元素都是使用逗點隔開，左右有中括號 []。

```python
# scores 是一個串列，裏面每個元素都是整數。
scores = [78, 29, 56, 80, 99]
print(scores)
```

## 索引值

我們可以使用索引值 (index) 來存取串列中的元素。

注意：索引值由 0 開始。

```python
print(scores[0]) # 印出第 1 個元素
print(scores[1]) # 印出第 2 個元素
```

## 串列內容的個數

len(a) 取得 a 這個串列中元素的個數。

len 是 length (長度) 的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1])
```
